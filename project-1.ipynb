```python
df.rename(columns = {'activitydate':'activity_date' , 'totalsteps': 'total_steps', 'totaldistance': 'total_distance', 'trackerdistance': 'tracker_distance',
       'loggedactivitiesdistance': 'logged_activities_distance', 'veryactivedistance': 'very_active_distance','moderatelyactivedistance': 'moderately_active_distance', 'lightactivedistance': 'light_active_distance','sedentaryactivedistance': 'sedentary_active_distance', 'veryactiveminutes': 'very_active_minutes', 'fairlyactiveminutes': 'fairly_active_minutes',
       'lightlyactiveminutes': 'lightly_active_minutes', 'sedentaryminutes': 'sedentary_minutes'}, inplace=True)
df.columns
```

```python
# create columns (datetime function is used)
day_of_week = df['activity_date'].dt.day_name()
df['day_of_week'] = day_of_week

df['n_day_of_week'] = df['activity_date'].dt.weekday #0 monday,6 sunday
```

```python
#subset the data
df = df[['id', 'activity_date', 'total_steps', 'total_distance',
       #'tracker_distance', 'logged_activities_distance',
       #'very_active_distance', 'moderately_active_distance',
       #'light_active_distance', 'sedentary_active_distance',
       'very_active_minutes', 'fairly_active_minutes',
       'lightly_active_minutes', 'sedentary_minutes', 'calories']].copy()
```

```python
df.head(3)
```

```python
#analysis
# Group by 'id' and calculate average total distance
# Define new conditions for classification based on Total Distance
# Assign categories based on conditions
# Create a mapping of ID to Distance Level
# Apply the classification back to the original dataframe

id_grp = df.groupby(['id'])
id_avg_dist = id_grp['total_distance'].mean().sort_values(ascending= False)
id_avg_dist = id_avg_dist.to_frame()

conditions = [
    (id_avg_dist <= 5),  # Low Distance
    (id_avg_dist > 5) & (id_avg_dist <= 10),  # Moderate Distance
    (id_avg_dist > 10)  # High Distance
]

values = ['low_distance', 'moderate_distance', 'high_distance']
id_avg_dist['distance_level'] = np.select(conditions, values)

id_distance_level = id_avg_dist['distance_level']

df['distance_level'] = [id_distance_level[c] for c in df['id']]
```

```python

#low use: less than 8 hours per day

#normal use: between 8 and 16 hours per day

#high use more than 16 hours and up to 24 hours
df.head(4)
```

```python
# Assuming you have a column for total active minutes per day
# Convert minutes to hours
# Group by 'id' and calculate average daily active hours
# Define conditions for classification based on active hours
# Assign categories based on conditions
# Create a mapping of ID to usage level
# Apply the classification back to the original dataframe

df['total_active_hours'] = (
    df['very_active_minutes'] + df['fairly_active_minutes'] + df['lightly_active_minutes']
) / 60  


id_grp = df.groupby(['id'])
id_avg_hours = id_grp['total_active_hours'].mean().sort_values(ascending=False)
id_avg_hours = id_avg_hours.to_frame()

conditions = [
    (id_avg_hours < 8),  # Low Use
    (id_avg_hours >= 8) & (id_avg_hours <= 16),  # Normal Use
    (id_avg_hours > 16) & (id_avg_hours <= 24)  # High Use
]

values = ['low_use', 'normal_use', 'high_use']
id_avg_hours['usage_level'] = np.select(conditions, values)

id_usage_level = id_avg_hours['usage_level']


df['usage_level'] = [id_usage_level[c] for c in df['id']]

df.head(4)


```
